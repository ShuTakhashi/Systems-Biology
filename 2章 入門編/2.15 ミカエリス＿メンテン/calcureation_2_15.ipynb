## ミカエリス=メンテンの式
断熱消去などで縮約
$$
\begin{align}
\frac{d[P]}{dt} &= k_p \frac{[E]_T [S]}{K_m+[S]} \\
\frac{d[S]}{dt}+\frac{d[ES]}{dt}+\frac{d[P]}{dt} &= 0\\
\frac{d[ES]}{dt} &= 0\\
\end{align}
$$

```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
t=0.0
dt=0.01
endTime=300.0
steps=endTime/dt
```

```python
S_0 = 10
ES_0 = 0
P_0 = 0
```

```python
E_t = 0.1
k1 = 1  #k-
k2 = 10 #k+
kp = 0.5
Km = (k1+kp)/k2
```

```python
t_array = np.arange(0,endTime,dt)
S_array = np.array([])
#ES_array = np.array([])
P_array = np.array([])

S = S_0
P = P_0
```

```python
for i in range(int(steps)):
    t+= dt
    S+= -dt*(kp*(E_t*S)/(Km+S))
    P+= dt*(kp*(E_t*S)/(Km+S))
    S_array = np.append(S_array,S)
    P_array = np.append(P_array,P)


#P_array = -S_array
```

```python
plt.plot(t_array, S_array,color='red')
plt.plot(t_array, P_array,color='blue')

plt.xlabel('t')
plt.ylabel('[S],[P]')
plt.show()
```

```python
S_array = np.array([])
P_array = np.array([])
ES_array = np.array([])

S = S_0
P = P_0
ES = ES_0
```

```python
for i in range(int(steps)):
    t+= dt
    S+= dt*(-k2*(E_t-ES)*S+k1*ES)
    ES+= dt*(k2*(E_t-ES)*S-k1*ES-kp*ES)
    P+= dt*(kp*ES)
    S_array = np.append(S_array,S)
    P_array = np.append(P_array,P)
    ES_array = np.append(ES_array,ES)
```

```python
plt.plot(t_array, S_array,color='red')
plt.plot(t_array, P_array,color='blue')

plt.xlabel('t')
plt.ylabel('[S],[P]')
plt.show()

plt.plot(t_array, ES_array,color='green')

plt.xlabel('t')
plt.ylabel('[ES]')
plt.show()
```
